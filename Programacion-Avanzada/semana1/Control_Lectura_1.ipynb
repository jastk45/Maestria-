## Complejidad computacional en operaciones sobre datos

Búsqueda en `list` frente a `dict`.

## 1-2. Estructuras y consultas

```python
import time

SIZES = [10_000, 50_000, 100_000]
N_CONSULTAS = 200


def construir(n):
    """Lista y dict con los ids 0..n-1, y 200 valores que no existen."""
    lista = list(range(n))
    dicc = {i: True for i in range(n)}
    consultas = [n + k for k in range(N_CONSULTAS)]
    return lista, dicc, consultas
```

## 3. Medición

Las estructuras se construyen antes del cronómetro: solo se mide la búsqueda.

```python
def medir(estructura, consultas):
    """Tiempo total en ms de buscar cada valor de consultas."""
    inicio = time.perf_counter()
    for v in consultas:
        _ = v in estructura
    return (time.perf_counter() - inicio) * 1000


resultados = []
for n in SIZES:
    lista, dicc, consultas = construir(n)      # fuera de la medicion
    resultados.append((n, medir(lista, consultas), medir(dicc, consultas)))
```

## 4. Resultados

```python
print(f"{'n':>8}{'list_ms':>12}{'dict_ms':>12}")
for n, list_ms, dict_ms in resultados:
    print(f"{n:>8}{list_ms:>12.4f}{dict_ms:>12.4f}")
```

## 5. Interpretación

**1. `list`.** El tiempo crece proporcional a `n`. El operador `in` recorre la lista elemento por elemento y, al no existir los valores buscados, la recorre completa.

**2. `dict`.** El tiempo se mantiene constante. Es una tabla hash: calcula el hash de la clave y accede directo a su posición.

**3. Notación.**

| Estructura | Búsqueda |
|---|---|
| `list` | **O(n)** |
| `dict` | **O(1)** en promedio |

*En promedio* porque las colisiones de hash pueden degradar la búsqueda.

**4. Qué elegir.** El `dict`. Con *k* búsquedas el costo es O(n + k) frente a O(n × k) de la lista: la construcción se paga una vez y cada consulta es constante.
